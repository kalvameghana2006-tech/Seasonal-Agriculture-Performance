### 12.1 Investigating Missing Numerical Values

The missing values identified during the data quality assessment are present in numerical variables.

The affected variables are:

- Rainfall
- Soil moisture
- Crop yield

Since these are numerical measurements, an appropriate numerical imputation method can be considered rather than removing complete observations.

```python
# Display summary statistics for variables containing missing values

df_clean[['Rainfall_mm', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha']].describe()
```

### 12.2 Mean and Median Comparison

The mean and median values of the affected variables are compared before selecting an imputation method.

The median is generally less sensitive to extreme observations than the mean. Therefore, if a variable shows substantial skewness or extreme values, median-based imputation may provide a more robust estimate for missing observations.

```python
# Compare mean and median for variables with missing values

comparison = pd.DataFrame({
    'Mean': df_clean[['Rainfall_mm', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha']].mean(),
    'Median': df_clean[['Rainfall_mm', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha']].median()
})

comparison
```

### 12.3 Distribution of Variables with Missing Values

The distributions of rainfall, soil moisture, and yield are visualized to understand their spread and identify whether extreme observations are present.

This supports the decision regarding the most appropriate method for handling missing values.

```python
# Visualize distributions of variables with missing values

for column in ['Rainfall_mm', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha']:

    plt.figure(figsize=(8, 5))

    sns.histplot(df_clean[column].dropna(), kde=True)

    plt.title(f'Distribution of {column}')
    plt.xlabel(column)
    plt.ylabel('Frequency')
    plt.tight_layout()

    plt.show()
```

### 12.4 Handling Missing Values

The missing values in the numerical variables are handled using median imputation.

Median imputation is selected because the median is less affected by extreme observations than the mean. This approach allows us to retain the available records while reducing the influence of unusually large observations on the imputed values.

The following variables are treated:

- `Rainfall_mm`
- `Soil_Moisture_pct`
- `Yield_Tonnes_Ha`

```python
# Fill missing values using the median of each respective column

columns_to_impute = [
    'Rainfall_mm',
    'Soil_Moisture_pct',
    'Yield_Tonnes_Ha'
]

for column in columns_to_impute:
    df_clean[column] = df_clean[column].fillna(df_clean[column].median())

print("Missing values have been handled successfully.")
```

### 12.5 Verification After Missing Value Treatment

After applying the missing-value treatment, the dataset is checked again to confirm that the affected variables no longer contain missing observations.


```python
# Check missing values after imputation

remaining_missing = df_clean.isnull().sum()

remaining_missing = remaining_missing[remaining_missing > 0]

if len(remaining_missing) == 0:
    print("No missing values remain in the dataset.")
else:
    print(remaining_missing)
```

### 12.6 Duplicate Verification

```python
# Check duplicate rows in the cleaned dataset

duplicate_count_clean = df_clean.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count_clean}")
```

## 10. Statistical Analysis

Statistical analysis is used to compare the distribution and variability of agricultural performance across seasons.

This section focuses on:
- Comparing mean and median yield across seasons
- Measuring variability using standard deviation and IQR
- Estimating 95% confidence intervals for seasonal yield
- Comparing seasonal profit distributions
- Understanding whether seasonal averages are consistent or highly variable

These descriptive statistical measures help provide deeper evidence for the seasonal patterns observed during exploratory analysis.

```python
seasonal_yield_stats = df_clean.groupby('Season')['Yield_Tonnes_Ha'].agg(
    Count='count',
    Mean='mean',
    Median='median',
    Std_Dev='std',
    Minimum='min',
    Maximum='max'
)

seasonal_yield_stats['Q1'] = df_clean.groupby('Season')['Yield_Tonnes_Ha'].quantile(0.25)
seasonal_yield_stats['Q3'] = df_clean.groupby('Season')['Yield_Tonnes_Ha'].quantile(0.75)

seasonal_yield_stats['IQR'] = (
    seasonal_yield_stats['Q3'] - seasonal_yield_stats['Q1']
)

seasonal_yield_stats.round(3)
```

### Interpretation of Seasonal Yield Statistics

The table provides a detailed statistical comparison of crop yield across Kharif, Rabi, and Zaid seasons.

Mean and median values describe the central tendency, while standard deviation and IQR describe the variability of yield within each season.

Comparing the mean and median also helps identify whether extreme yield values may be influencing the seasonal averages.


```python
yield_ci = df_clean.groupby('Season')['Yield_Tonnes_Ha'].agg(
    Count='count',
    Mean='mean',
    Std_Dev='std'
)

yield_ci['Standard_Error'] = (
    yield_ci['Std_Dev'] / np.sqrt(yield_ci['Count'])
)

yield_ci['CI_Lower'] = (
    yield_ci['Mean'] - 1.96 * yield_ci['Standard_Error']
)

yield_ci['CI_Upper'] = (
    yield_ci['Mean'] + 1.96 * yield_ci['Standard_Error']
)

yield_ci.round(3)
```

### 95% Confidence Interval for Seasonal Yield

A 95% confidence interval provides an estimated range around the seasonal mean yield.

The interval is calculated using:

**Mean ± 1.96 × Standard Error**

A narrower interval indicates that the estimated seasonal mean is more precise, while a wider interval indicates greater uncertainty around the estimated mean.

```python
plt.figure(figsize=(8, 5))

x = np.arange(len(yield_ci))

plt.errorbar(
    x,
    yield_ci['Mean'],
    yerr=1.96 * yield_ci['Standard_Error'],
    fmt='o',
    capsize=5
)

plt.xticks(x, yield_ci.index)
plt.xlabel('Season')
plt.ylabel('Mean Yield (Tonnes/Ha)')
plt.title('Seasonal Mean Yield with 95% Confidence Intervals')

plt.tight_layout()
plt.show()
```

### Interpretation

The confidence interval plot shows the estimated mean yield for each season together with its uncertainty range.

This provides statistical support for comparing seasonal yield performance rather than relying only on the difference between the average values.

## 11. Outlier Analysis

Outlier analysis is performed to identify observations that are unusually high or low compared with the majority of the data.

In agricultural datasets, extreme values may represent genuine differences in farm size, production, resource usage, or economic performance. Therefore, an identified outlier should not automatically be treated as an error.

The analysis will:
- Identify outliers using the IQR method
- Count outliers in important numerical variables
- Compare the percentage of outliers across variables
- Visualize distributions using boxplots
- Examine extreme observations
- Decide whether the identified values should be retained

```python
def count_iqr_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    return len(outliers), lower_bound, upper_bound


outlier_columns = [
    'Rainfall_mm',
    'Yield_Tonnes_Ha',
    'Production_Tonnes',
    'Revenue_INR',
    'Total_Cost_INR',
    'Profit_INR',
    'Water_Used_m3',
    'Water_Efficiency_t_per_1000m3'
]

outlier_results = []

for column in outlier_columns:
    count, lower, upper = count_iqr_outliers(df_clean, column)

    outlier_results.append([
        column,
        count,
        lower,
        upper
    ])

outlier_table = pd.DataFrame(
    outlier_results,
    columns=[
        'Variable',
        'Outlier_Count',
        'Lower_Bound',
        'Upper_Bound'
    ]
)

outlier_table.round(2)
```

### IQR Method

The Interquartile Range (IQR) method identifies observations outside the following limits:

**Lower Bound = Q1 − 1.5 × IQR**

**Upper Bound = Q3 + 1.5 × IQR**

where Q1 is the first quartile and Q3 is the third quartile.

Observations outside these limits are flagged as potential outliers.

```python
outlier_table['Outlier_Percentage'] = (
    outlier_table['Outlier_Count'] /
    len(df_clean)
) * 100

outlier_table[
    ['Variable', 'Outlier_Count', 'Outlier_Percentage']
].sort_values(
    'Outlier_Percentage',
    ascending=False
).round(2)
```

### Interpretation of Outlier Counts

The outlier percentages show how frequently unusually high or low observations occur in each variable.

Variables with a higher number of flagged observations require closer examination.

However, the IQR method identifies statistical extremes rather than data-entry errors. Therefore, the presence of an outlier does not by itself justify removing the observation.
